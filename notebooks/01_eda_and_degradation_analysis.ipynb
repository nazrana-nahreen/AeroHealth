# ✈️ AeroHealth AI: Exploratory Data Analysis & Degradation Modeling
This notebook analyzes the **NASA C-MAPSS Turbofan Engine Degradation Simulation (FD001)** dataset to explore sensor drift, degradation trajectories, and Remaining Useful Life (RUL) modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path

# Add project root to sys.path
sys.path.append('..')
from src.data_loader import prepare_dataset
from src.config import SENSOR_NAMES, FD001_DROP_SENSORS

# Ingest dataset
train_df, test_df, true_rul = prepare_dataset('FD001')
print(f'Train shape: {train_df.shape}, Test shape: {test_df.shape}')
```

### 1. Engine Operating Life Distribution (Run-to-Failure)

```python
engine_lifespans = train_df.groupby('unit_number')['time_in_cycles'].max()
print(f'Shortest lifespan: {engine_lifespans.min()} cycles')
print(f'Longest lifespan:  {engine_lifespans.max()} cycles')
print(f'Median lifespan:   {engine_lifespans.median()} cycles')

plt.figure(figsize=(9, 4))
plt.hist(engine_lifespans, bins=20, color='royalblue', edgecolor='black')
plt.title('Distribution of Engine Lifespans in Training Fleet')
plt.xlabel('Operating Cycles to Failure')
plt.ylabel('Number of Engines')
plt.grid(True, alpha=0.3)
plt.show()
```

### 2. Sensor Degradation Trajectory (Engine #1)
Observe how key thermal and mechanical sensors drift as the engine approaches failure.

```python
unit_1 = train_df[train_df['unit_number'] == 1]
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

key_sensors = ['sensor_2', 'sensor_3', 'sensor_4', 'sensor_7']
for ax, s in zip(axes.flat, key_sensors):
    ax.plot(unit_1['time_in_cycles'], unit_1[s], color='firebrick', lw=1.5)
    ax.set_title(f"{s}: {SENSOR_NAMES.get(s, s)}")
    ax.set_xlabel('Cycles')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
